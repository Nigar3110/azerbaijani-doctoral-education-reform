```python
import pandas as pd

# Read CSV file
df = pd.read_csv("../data/supervisor_survey.csv")

# Show first rows
print(df.head())
```

```python
df.rename(columns={"Elmi tədqiqatın ixtisas sahəsi  ": "subject"}, inplace = True)
```

```python
df = df.fillna(0)
```

```python
def categorize(field):
    if pd.isna(field) or field == 0:
        return "Unknown"
    
    field = str(field).lower()
    
    if any(x in field for x in ["engineering", "oil", "petroleum", "mechanical", "transport", "telecommunication", "technology"]):
        return "Engineering & Technology"
    
    elif any(x in field for x in ["physics", "chemistry", "biology", "microbiology", "ecology", "geography", "earth", "genetics"]):
        return "Natural Sciences"
    
    elif any(x in field for x in ["computer", "information", "ai", "informatics", "security", "system analysis"]):
        return "Computer Science & IT"
    
    elif any(x in field for x in ["economics", "finance", "marketing", "economy"]):
        return "Economics & Business"
    
    elif any(x in field for x in ["history", "philology", "philosophy", "linguistics", "culturology"]):
        return "Humanities"
    
    elif any(x in field for x in ["sociology", "political", "psychology", "pedagogy", "relations"]):
        return "Social Sciences"
    
    else:
        return "Interdisciplinary"

df["Category"] = df["subject"].apply(categorize)
```

```python
counts = df["Category"].value_counts()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,7))

plt.pie(
    counts,
    labels=counts.index,
    autopct='%1.1f%%',   # shows percentages
    startangle=90
)

plt.title("Distribution of Categories")
plt.show()
```

```python
df.rename(columns={"Vəzifəniz (ştat və ya yarımştat)": "position"}, inplace = True)
```

```python
def categorize_position(text):
    if pd.isna(text) or text == 0:
        return "Unknown"
    
    text = str(text).lower()
    
    # Professor
    if "professor" in text:
        return "Professor"
    
    # Associate professor
    elif "dosent" in text:
        return "Associate Professor"
    
    # Researcher
    elif any(x in text for x in ["elmi işçi", "elmi isci", "researcher"]):
        return "Researcher"
    
    # Administrative
    elif any(x in text for x in ["müdiri", "mudiri", "direktor", "rəhbər", "katib"]):
        return "Administrative / Leadership"
    
    # Lecturer
    elif any(x in text for x in ["müəllim", "muellim"]):
        return "Lecturer"
    
    # Technical
    elif any(x in text for x in ["mühəndis", "expert"]):
        return "Technical / Other"
    
    else:
        return "Unknown"

df["Position_Category"] = df["position"].apply(categorize_position)
```

```python
data_cluster = data_cluster.dropna()
```

```python
X = data_cluster[['FinancialSatisfaction', 'stress', 'support']]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)  # try 3–5
labels = kmeans.fit_predict(X_scaled)

data_cluster['cluster'] = labels
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, labels)
print("Silhouette Score:", score)
```

```python
ct = pd.crosstab(data_cluster['cluster'], data_cluster['Position_Category'], normalize='index')
print(ct)
```

```python
ct = pd.crosstab(
    data_cluster['cluster'],
    data_cluster['Position_Category'],
    normalize='index'
)

plt.figure(figsize=(10,5))
sns.heatmap(ct, annot=True, cmap="YlGnBu")

plt.title("Position Distribution per Cluster")
plt.xlabel("Position Category")
plt.ylabel("Cluster")

plt.show()
```

```python
chi2, p, dof, expected = chi2_contingency(ct)

print("Chi-square:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
```
